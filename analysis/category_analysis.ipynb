# 샘플로 테스트

```python
import pandas as pd
import numpy as np

# 감정군 목록
all_emotions = ["love", "hope", "sadness", "fear", "disgust", "anger"]

# 데이터 불러오기
df = pd.read_pickle('../results/emotion_analysis.pkl')
sample_df = df.sample(1000, random_state=42)
sample_df.head()
```

```python
emotion_count_cols = [
    'count_anger', 'count_hope', 'count_love', 'count_sadness', 'count_fear', 'count_disgust'
]

emotion_ratio_cols = [
    'ratio_anger', 'ratio_hope', 'ratio_love', 'ratio_sadness', 'ratio_fear', 'ratio_disgust'
]
```

## 장르별 분석

```python
# Unknown 제외 데이터셋
df_genre = sample_df[sample_df['genre'] != 'Unknown']
```

```python
# 장르별 감정군 Count / 비율
genre_emotion_count = df_genre.groupby('genre')[emotion_count_cols].sum()
genre_emotion_ratio = genre_emotion_count.div(genre_emotion_count.sum(axis=1), axis=0)

display(genre_emotion_count)
display(genre_emotion_ratio)
```

```python
# 장르별 평균 emotion_score / normalized_emotion_score
genre_emotion_score = df_genre.groupby('genre')[['emotion_score', 'normalized_emotion_score']].mean()

display(genre_emotion_score)
```

```python
# emotion_words 컬럼 문자열 split → explode 처리
sample_explode = df_genre.assign(emotion_words=df_genre['emotion_words'].str.split(', ')).explode('emotion_words')

# 장르별 가장 많이 등장한 감정 단어 Top-10
genre_top10_words = (
    sample_explode.groupby('genre')['emotion_words']
    .value_counts()
    .groupby(level=0)
    .nlargest(10)
    .reset_index(level=0, drop=True)
)

display(genre_top10_words)
```

## 시대별 분석

```python
# 연도를 10년 단위로 묶기 (Decade 컬럼 생성)
sample_df['decade'] = (sample_df['year'] // 10) * 10
```

```python
# 시대별 감정군 Count / 비율
decade_emotion_count = sample_df.groupby('decade')[emotion_count_cols].sum()
decade_emotion_ratio = decade_emotion_count.div(decade_emotion_count.sum(axis=1), axis=0)

display(decade_emotion_count)
display(decade_emotion_ratio)
```

```python
# 시대대별 평균 점수 변화
decade_emotion_score = sample_df.groupby('decade')[['emotion_score', 'normalized_emotion_score']].mean()

display(decade_emotion_score)
```

```python
# emotion_words 컬럼 처리
df_explode = sample_df.assign(emotion_words=sample_df['emotion_words'].str.split(', ')).explode('emotion_words')

# 시대별 감정 단어 Top-N
top_n = 10
decade_top_words = (
    df_explode.groupby('decade')['emotion_words']
    .value_counts()
    .groupby(level=0)
    .nlargest(top_n)
    .reset_index(level=0, drop=True)
)

display(decade_top_words)
```

## 아티스트별 분석

```python
# 상위 N명 아티스트 (곡 수 기준)
top_n = 20
top_artists = sample_df['artist'].value_counts().head(top_n).index.tolist()
df_top_artists = sample_df[sample_df['artist'].isin(top_artists)]
```

```python
# 감정군별 평균 Count / 비율
artist_emotion_count = df_top_artists.groupby('artist')[emotion_count_cols].mean()
artist_emotion_ratio = artist_emotion_count.div(artist_emotion_count.sum(axis=1), axis=0)

# 아티스트별 평균 감정 점수
artist_emotion_score = df_top_artists.groupby('artist')[['emotion_score', 'normalized_emotion_score']].mean()

display(artist_emotion_count)
display(artist_emotion_ratio)
display(artist_emotion_score)
```

```python
# 상위 N명 아티스트만 대상으로 처리
top_artists = sample_df['artist'].value_counts().head(20).index.tolist()
df_top_artists = sample_df[sample_df['artist'].isin(top_artists)]

df_explode_artist = df_top_artists.assign(emotion_words=df_top_artists['emotion_words'].str.split(', ')).explode('emotion_words')

# 아티스트별 감정 단어 Top-N
top_n = 10
artist_top_words = (
    df_explode_artist.groupby('artist')['emotion_words']
    .value_counts()
    .groupby(level=0)
    .nlargest(top_n)
    .reset_index(level=0, drop=True)
)

display(artist_top_words)
```

```python
# 특정 아티스트 감정 변화 추이
target_artist = 'Taylor Swift'
artist_df = sample_df[sample_df['artist'] == target_artist]

artist_decade_emotion = artist_df.groupby('decade')[emotion_count_cols].sum()
artist_decade_score = artist_df.groupby('decade')[['emotion_score', 'normalized_emotion_score']].mean()

display(artist_decade_emotion)
display(artist_decade_score)
```
